## Single Perceptron

```python
# Binary Cross-Entropy Loss for scalar
def binary_cross_entropy_loss(prediction, target):
    epsilon = 1e-8  # To prevent log(0)
    prediction = torch.clamp(prediction, epsilon, 1 - epsilon)
    return -(target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
```

```python
x = torch.tensor(6.7) #input data
y = torch.tensor(0.0) #output data

w = torch.tensor(1.0, requires_grad=True) #init random weight 
b = torch.tensor(0.0, requires_grad=True) #init random bias

for i in range(0,2):
    #Forward
    z = w * x + b # linear operation
    print(f"z",z)

    y_pred = torch.sigmoid(z) # activation
    print(f"y_pred",y_pred)

    loss = binary_cross_entropy_loss(y_pred, y) #calculate loss
    print(f"loss",loss)

    #backword
    loss.backward()
    print(w.grad) #updated w
    print(b.grad) #updated b

    w.grad.zero_() #remove previous w
    b.grad.zero_() #remove previous b
    print(f"=========================")

```
